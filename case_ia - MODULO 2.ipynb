## **O que é um modelo preditivo?**

Um modelo preditivo é um tipo de modelo de machine learning que tem como objetivo principal fazer previsões sobre novos dados com base em padrões aprendidos a partir de dados históricos.

---

Esses modelos utilizam padrões extraídos dos dados para fazer previsões sobre novas observações. Em termos formais, esses modelos buscam aprender uma função $$f: x_{i} \in X \rightarrow Y$$ que relacione os vetores de características $x_{i} \in X$ com a variável alvo $Y$, de forma que seja possível estimar um $Y$ para um novo $x_{i} \in X$ não observado anteriormente.

---

Os modelos preditivos podem ser utilizados para diferentes tipos de tarefas, como prever se um cliente pagará um empréstimo (classificação), estimar o valor de um imóvel (regressão) ou antecipar o comportamento de uma série temporal (como preços de ativos financeiros).

No entanto, as abordagens para construir esse tipo de modelo variam bastante, dependendo da forma como os dados são tratados e da estratégia utilizada para ajustar os parâmetros do modelo. Podemos dividir os modelos preditivos em algumas categorias principais, são elas:



### **1. Modelos Estatísticos e Probabilísticos**

Esses modelos assumem que os dados seguem uma distribuição probabilística específica e ajustam seus parâmetros para maximizar a verossimilhança dos dados observados.

✅ **Objetivo:** Estimar a distribuição dos dados e inferir a relação entre as variáveis.

**Exemplos:**

- Regressão Linear: Assume que a relação entre as variáveis é linear e que os resíduos seguem uma distribuição normal.

- Regressão Logística: Usa a função sigmoide para modelar probabilidades em problemas de classificação binária.

- Naïve Bayes: Baseia-se na regra de Bayes e assume independência condicional entre os atributos.

📌 **Diferencial:** Esses modelos são baseados em conceitos estatísticos bem definidos e geralmente interpretáveis.

### **2. Modelos Baseados em Otimização**

Esses modelos aprendem a função $f$ otimizando uma métrica específica, sem necessariamente assumir uma distribuição subjacente dos dados.

✅ **Objetivo:** Encontrar um modelo que minimize uma função de erro ou maximize uma margem de separação.

**Exemplos:**

- SVM (Máquinas de Vetores de Suporte): Maximiza a margem entre as classes em problemas de classificação.

- Redes Neurais (MLPs clássicas): Ajustam pesos para minimizar um erro através do gradiente descendente.

📌 **Diferencial:** Baseiam-se em técnicas numéricas de otimização e não exigem uma suposição explícita sobre a distribuição dos dados.

### **3. Modelos Baseados em Particionamento de Dados**

Esses modelos criam regras para dividir os dados em subconjuntos homogêneos, organizando-os hierarquicamente.

✅ **Objetivo:** Criar regras de decisão que segmentem os dados em diferentes categorias.

**Exemplos:**

- Árvores de Decisão: Criam um conjunto de regras sequenciais que levam a decisões.

- Random Forests: Conjunto de árvores de decisão para aumentar a robustez.

📌 **Diferencial:** São fáceis de interpretar e não exigem transformações matemáticas complexas.

### **4. Modelos Baseados em Agrupamento (Clustering)**

Modelos que tentam encontrar padrões e estruturas nos dados sem rótulos supervisionados.

✅ **Objetivo:** Agrupar os dados de forma a maximizar a semelhança dentro dos clusters e minimizar entre os clusters.

**Exemplos:**

- K-Means: Baseado na minimização da soma das distâncias dos pontos ao centro do cluster.

- DBSCAN: Detecta clusters com base na densidade dos dados.

📌 **Diferencial:** Úteis para análise exploratória e segmentação de dados sem supervisão.



### **5. Modelos Baseados em Instâncias**

Esses modelos fazem previsões baseadas nos exemplos mais próximos no conjunto de treinamento, sem construir explicitamente uma função generalizada durante o aprendizado.

✅ **Objetivo:** Utilizar diretamente os dados observados para fazer previsões sem uma fase explícita de aprendizado de parâmetros.

**Exemplo:**
    
- KNN (K-Nearest Neighbors): Classifica ou estima valores com base nos $k$ vizinhos mais próximos.

📌 Diferencial: São conhecidos como modelos preguiçosos (lazy learners) porque não possuem uma fase explícita de treinamento; todo o esforço computacional ocorre no momento da previsão.

### **A importância da amostra de treinamento para um modelo preditivo**

No fim das contas, modelos preditivos tentam encontrar uma função matemática que capture a relação entre as variáveis de entrada (características) e a variável de saída (o que queremos prever). Essa função pode ser vista como uma aproximação da verdadeira relação existente nos dados. No entanto, os dados usados para treinar o modelo representam apenas uma amostra de um conjunto muito maior e desconhecido. Se essa amostra for pequena, enviesada ou não representativa, o modelo pode aprender padrões artificiais que não se repetem em novos dados — um problema conhecido como overfitting.

Diferentes modelos fazem suposições distintas sobre essa relação. Modelos estatísticos assumem que os dados seguem certas distribuições probabilísticas, enquanto modelos baseados em otimização simplesmente ajustam funções para melhor se encaixar nos exemplos observados, sem pressupor uma estrutura fixa. Em ambos os casos, a qualidade da amostra influencia diretamente a capacidade do modelo de fazer boas previsões.

Por isso, ao treinar um modelo, é essencial garantir que os dados de entrada representem bem o fenômeno que estamos tentando modelar. Métodos como validação cruzada, regularização e aumento da diversidade dos dados ajudam a reduzir vieses e melhorar a capacidade do modelo de generalizar para novas situações.

## **Tradeoff entre Bias e Variância** 

Ao se construir um modelo de machine learning, seja qual for o objetivo, é necessário levar em consideração possíveis erros de previsão que podem interferir com a acurácia do resultado final. 

Duas fontes desses erros de previsão são o Bias (ou viés em português) e a Variância, que servem como uma maneira de medir e otimizar o desempenho do nosso modelo.

### **Explicando o Bias**

Resumidamente, o Bias nada mais é do que a medida da distância entre o resultado esperado/real das nossas previsões e o resultado gerado pelo modelo. 

- Como exemplo, podemos dizer que modelos de machine learning com elevado viés tendem a fazer suposições fortes sobre a forma dos dados e causar subajuste (underfitting). Já um modelo excessivamente simplista tende a ter alto viés e baixa variância, tendendo a ter altos erros de treinamento e altos erros de previsão.  

### **Explicando a Variância**

A variância diz respeito à variabilidade e consistência dos resultados gerados pelo modelo, medindo o quanto as previsões mudam com diferentes conjuntos de dados de treinamento. Dessa forma, ao se repetir novamente o treinamento, é medido o quanto as previsões do modelo variam.
- Um modelo com alta variância, por exemplo, é sensível a ruídos nos dados de treinamento e causa sobreajuste (overfitting) 

### **Explicando o Tradeoff entre Bias e Variância**

Para que o modelo seja robusto e de alto desempenho preditivo, é fundamental que haja um equilíbrio entre o Bias e a Variância, em que ambos sejam reduzidos. No entanto, algumas outras situações podem acontecer, como:

- Baixo Bias e Alta Variância: O modelo "decora" o dataset de treinamento e tem um desempenho ruim em dados não vistos, pois não consegue generalizar (overfitting)

- Alto Bias e Baixa Variância: Como o modelo é excessivamente simplificado, ele falha em acertar as previsões, tanto para os dados que já viu quanto para dados novos, mas apresenta previsões consistentes.

- Alto Viés e Alta Variância: O modelo é inconsistente, e possui tanto uma baixa acurácia quanto uma baixa precisão.



É importante salientar que o número de parâmetros presente no modelo e, consequentemente, sua complexidade, afetam diretamente esse equilíbrio necessário para a construção de um bom modelo. 

Como se vê na figura abaixo, na medida que o modelo se torna mais complexo e possui mais parâmetros, a variabilidade nos valores previstos no conjunto de teste aumenta, resultando em alta variância, enquanto que o Bias é reduzido constantemente. 

Portanto, ao construirmos um modelo, buscamos simultaneamente equilibrar o Bias e a variância para atingir o desempenho ideal do modelo, o que não apenas gera bons resultados no treinamento, mas também generaliza bem para dados de teste novos.

Alguns exemplos de modelos conhecidos com:
-   Alta variância:
    - Árvores de decisão sem poda.
    - Regressão polinomial com altos graus.
    - KNN com k muito baixo.
    
-   Alto Bias:
    - KNN com k muito alto
    - Regressão linear simples ou múltipla

### Mas então, como podemos ajustar essas medidas da melhor forma possível?

O tradeoff entre Bias e Variância vai sempre depender do algoritmo a ser utilizado para a construção do modelo. 

Por exemplo:

- O algoritmo de KNN, tem baixo Bias e alta Variância, mas o tradeoff pode ser alterado ao aumentar o valor de k, que aumenta o Bias do modelo.

- O algoritmo de Support Vector Machine (SVM) tem baixo Bias e alta Variância, mas a compensação pode ser alterada aumentando o parâmetro C, que influencia o número de violações da margem permitida nos dados de treinamento, o que aumenta a Viés, mas diminui a Variância.


Nesse sentido, encontrar esse equilíbrio é um processo trabalhoso e iterativo, sendo necessário treinar o modelo diversas vezes com variadas combinações de features, hiperparâmetros e com diferentes conjuntos de dados para treinamento.

## **Validação Cruzada** 

Uma técnica para avaliar a capacidade de generalização do modelo e, consequentemente, lidar da melhor forma com problemas como o overfitting é a validação cruzada (Cross-validation). 

- A validação cruzada é uma técnica que avalia um modelo dividindo os dados em múltiplos subconjuntos, treinando o modelo em um subconjunto e testando-o nos restantes de forma iterativa. Isso tem como intuito avaliar como o modelo se comporta com dados inéditos, simulando uma situação de uso real.

Uma das formas mais conhecidas de aplicar uma validação cruzada em um modelo é por meio da técnica de _k-fold cross-validation_. Nela, o conjunto de dados é dividido de forma aleatória em "K" folds (ou "dobras" em português)

O fluxo metodológico é simples, observe o exemplo abaixo.

```python
# IMPORTS
import pandas as pd
import numpy as np
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import MinMaxScaler, LabelEncoder
from sklearn.model_selection import cross_val_score
import warnings
from sklearn.exceptions import ConvergenceWarning

# =========================================
# 1. GERANDO UM DATASET SINTÉTICO
# =========================================

np.random.seed(42)

# Criando 1000 amostras
n = 1000

dataset = pd.DataFrame({
    'idade': np.random.randint(18, 70, n),
    'renda': np.random.randint(1000, 10000, n),
    'tempo_emprego': np.random.randint(0, 40, n),
    'checking_status': np.random.choice(['baixo', 'médio', 'alto'], n),
    'credit_history': np.random.choice(['bom', 'ruim', 'neutro'], n),
    'purpose': np.random.choice(['carro', 'educação', 'reforma', 'outros'], n),
    'savings_status': np.random.choice(['baixo', 'médio', 'alto'], n),
    'class': np.random.choice(['bom_pagador', 'mau_pagador'], n)
})

# =========================================
# 2. PRÉ-PROCESSAMENTO
# =========================================

# Dummies para variáveis categóricas
dataset_dummies = pd.get_dummies(dataset, 
    columns=['checking_status', 'credit_history', 'purpose', 'savings_status'], 
    drop_first=True
)

# Transformando a variável alvo em numérica
labelencoder = LabelEncoder()
dataset_dummies['class'] = labelencoder.fit_transform(dataset_dummies['class'])

# Separando X e y
X = dataset_dummies.drop(columns=['class'])
y = dataset_dummies['class']

# Normalização
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# =========================================
# 3. DEFINIÇÃO DO MODELO
# =========================================

model = MLPClassifier(hidden_layer_sizes=(25, 25),
                      activation='relu',
                      solver='adam',
                      max_iter=200,
                      random_state=42)

# =========================================
# 4. CROSS-VALIDATION
# =========================================

scores = cross_val_score(model, X_scaled, y, cv=10, scoring='accuracy')

print("Scores individuais:", np.round(scores, 3))
print("Média de acurácia: %.2f%%" % (scores.mean() * 100))
print("Desvio padrão: %.2f%%" % (scores.std() * 100))
```

Nesse exemplo, o conjunto de dados foi dividido em 10 "dobras". 
- No primeiro teste, a dobra 1 é o teste e as dobras 2, 3, 4, 5 são treino.
- No segundo teste, a dobra 2 é o teste e as dobras 1, 3, 4, 5 são treino, e sucessivamente.

Esse processo continua, e as estatísticas são coletadas em cada etapa.

No fim, é retornada uma lista com o resultado de cada um dos testes, para cada um dos subconjuntos, utilizando a métrica padrão do modelo: acurácia, para modelos de classificação, e R2 para modelos de regressão.

### 💡 Dica:

Recomendamos que utilizem para K os valores 5 e 10, pois estes não costumam causar erros de teste altos nem de Bias e nem de Variância.
