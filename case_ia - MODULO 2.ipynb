# **2. Fundamentos de Machine Learning**

## **O que é um modelo preditivo?**

Um modelo preditivo é um tipo de modelo de machine learning que tem como objetivo principal fazer previsões sobre novos dados com base em padrões aprendidos a partir de dados históricos.

---

Esses modelos utilizam padrões extraídos dos dados para fazer previsões sobre novas observações. Em termos formais, esses modelos buscam aprender uma função $$f: x_{i} \in X \rightarrow Y$$ que relacione os vetores de características $x_{i} \in X$ com a variável alvo $Y$, de forma que seja possível estimar um $Y$ para um novo $x_{i} \in X$ não observado anteriormente.

---

Os modelos preditivos podem ser utilizados para diferentes tipos de tarefas, como prever se um cliente pagará um empréstimo (classificação), estimar o valor de um imóvel (regressão) ou antecipar o comportamento de uma série temporal (como preços de ativos financeiros).

No entanto, as abordagens para construir esse tipo de modelo variam bastante, dependendo da forma como os dados são tratados e da estratégia utilizada para ajustar os parâmetros do modelo. Podemos dividir os modelos preditivos em algumas categorias principais, são elas:



### **1. Modelos Estatísticos e Probabilísticos**

Esses modelos assumem que os dados seguem uma distribuição probabilística específica e ajustam seus parâmetros para maximizar a verossimilhança dos dados observados.

✅ **Objetivo:** Estimar a distribuição dos dados e inferir a relação entre as variáveis.

**Exemplos:**

- Regressão Linear: Assume que a relação entre as variáveis é linear e que os resíduos seguem uma distribuição normal.

- Regressão Logística: Usa a função sigmoide para modelar probabilidades em problemas de classificação binária.

- Naïve Bayes: Baseia-se na regra de Bayes e assume independência condicional entre os atributos.

📌 **Diferencial:** Esses modelos são baseados em conceitos estatísticos bem definidos e geralmente interpretáveis.

### **2. Modelos Baseados em Otimização**

Esses modelos aprendem a função $f$ otimizando uma métrica específica, sem necessariamente assumir uma distribuição subjacente dos dados.

✅ **Objetivo:** Encontrar um modelo que minimize uma função de erro ou maximize uma margem de separação.

**Exemplos:**

- SVM (Máquinas de Vetores de Suporte): Maximiza a margem entre as classes em problemas de classificação.

- Redes Neurais (MLPs clássicas): Ajustam pesos para minimizar um erro através do gradiente descendente.

📌 **Diferencial:** Baseiam-se em técnicas numéricas de otimização e não exigem uma suposição explícita sobre a distribuição dos dados.

### **3. Modelos Baseados em Particionamento de Dados**

Esses modelos criam regras para dividir os dados em subconjuntos homogêneos, organizando-os hierarquicamente.

✅ **Objetivo:** Criar regras de decisão que segmentem os dados em diferentes categorias.

**Exemplos:**

- Árvores de Decisão: Criam um conjunto de regras sequenciais que levam a decisões.

- Random Forests: Conjunto de árvores de decisão para aumentar a robustez.

📌 **Diferencial:** São fáceis de interpretar e não exigem transformações matemáticas complexas.

### **4. Modelos Baseados em Agrupamento (Clustering)**

Modelos que tentam encontrar padrões e estruturas nos dados sem rótulos supervisionados.

✅ **Objetivo:** Agrupar os dados de forma a maximizar a semelhança dentro dos clusters e minimizar entre os clusters.

**Exemplos:**

- K-Means: Baseado na minimização da soma das distâncias dos pontos ao centro do cluster.

- DBSCAN: Detecta clusters com base na densidade dos dados.

📌 **Diferencial:** Úteis para análise exploratória e segmentação de dados sem supervisão.



### **5. Modelos Baseados em Instâncias**

Esses modelos fazem previsões baseadas nos exemplos mais próximos no conjunto de treinamento, sem construir explicitamente uma função generalizada durante o aprendizado.

✅ **Objetivo:** Utilizar diretamente os dados observados para fazer previsões sem uma fase explícita de aprendizado de parâmetros.

**Exemplo:**
    
- KNN (K-Nearest Neighbors): Classifica ou estima valores com base nos $k$ vizinhos mais próximos.

📌 Diferencial: São conhecidos como modelos preguiçosos (lazy learners) porque não possuem uma fase explícita de treinamento; todo o esforço computacional ocorre no momento da previsão.

### **A importância da amostra de treinamento para um modelo preditivo**

No fim das contas, modelos preditivos tentam encontrar uma função matemática que capture a relação entre as variáveis de entrada (características) e a variável de saída (o que queremos prever). Essa função pode ser vista como uma aproximação da verdadeira relação existente nos dados. No entanto, os dados usados para treinar o modelo representam apenas uma amostra de um conjunto muito maior e desconhecido. Se essa amostra for pequena, enviesada ou não representativa, o modelo pode aprender padrões artificiais que não se repetem em novos dados — um problema conhecido como overfitting.

Diferentes modelos fazem suposições distintas sobre essa relação. Modelos estatísticos assumem que os dados seguem certas distribuições probabilísticas, enquanto modelos baseados em otimização simplesmente ajustam funções para melhor se encaixar nos exemplos observados, sem pressupor uma estrutura fixa. Em ambos os casos, a qualidade da amostra influencia diretamente a capacidade do modelo de fazer boas previsões.

Por isso, ao treinar um modelo, é essencial garantir que os dados de entrada representem bem o fenômeno que estamos tentando modelar. Métodos como validação cruzada, regularização e aumento da diversidade dos dados ajudam a reduzir vieses e melhorar a capacidade do modelo de generalizar para novas situações.

## **Tradeoff entre Bias e Variância** 

Ao se construir um modelo de machine learning, seja qual for o objetivo, é necessário levar em consideração possíveis erros de previsão que podem interferir com a acurácia do resultado final. 

Duas fontes desses erros de previsão são o Bias (ou viés em português) e a Variância, que servem como uma maneira de medir e otimizar o desempenho do nosso modelo.

### **Explicando o Bias**

Resumidamente, o Bias nada mais é do que a medida da distância entre o resultado esperado/real das nossas previsões e o resultado gerado pelo modelo. 

- Como exemplo, podemos dizer que modelos de machine learning com elevado viés tendem a fazer suposições fortes sobre a forma dos dados e causar subajuste (underfitting). Já um modelo excessivamente simplista tende a ter alto viés e baixa variância, tendendo a ter altos erros de treinamento e altos erros de previsão.  

### **Explicando a Variância**

A variância diz respeito à variabilidade e consistência dos resultados gerados pelo modelo, medindo o quanto as previsões mudam com diferentes conjuntos de dados de treinamento. Dessa forma, ao se repetir novamente o treinamento, é medido o quanto as previsões do modelo variam.
- Um modelo com alta variância, por exemplo, é sensível a ruídos nos dados de treinamento e causa sobreajuste (overfitting) 

### **Explicando o Tradeoff entre Bias e Variância**

Para que o modelo seja robusto e de alto desempenho preditivo, é fundamental que haja um equilíbrio entre o Bias e a Variância, em que ambos sejam reduzidos. No entanto, algumas outras situações podem acontecer, como:

- Baixo Bias e Alta Variância: O modelo "decora" o dataset de treinamento e tem um desempenho ruim em dados não vistos, pois não consegue generalizar (overfitting)

- Alto Bias e Baixa Variância: Como o modelo é excessivamente simplificado, ele falha em acertar as previsões, tanto para os dados que já viu quanto para dados novos, mas apresenta previsões consistentes.

- Alto Viés e Alta Variância: O modelo é inconsistente, e possui tanto uma baixa acurácia quanto uma baixa precisão.



É importante salientar que o número de parâmetros presente no modelo e, consequentemente, sua complexidade, afetam diretamente esse equilíbrio necessário para a construção de um bom modelo. 

Como se vê na figura abaixo, na medida que o modelo se torna mais complexo e possui mais parâmetros, a variabilidade nos valores previstos no conjunto de teste aumenta, resultando em alta variância, enquanto que o Bias é reduzido constantemente. 

Portanto, ao construirmos um modelo, buscamos simultaneamente equilibrar o Bias e a variância para atingir o desempenho ideal do modelo, o que não apenas gera bons resultados no treinamento, mas também generaliza bem para dados de teste novos.

Alguns exemplos de modelos conhecidos com:
-   Alta variância:
    - Árvores de decisão sem poda.
    - Regressão polinomial com altos graus.
    - KNN com k muito baixo.
    
-   Alto Bias:
    - KNN com k muito alto
    - Regressão linear simples ou múltipla

### Mas então, como podemos ajustar essas medidas da melhor forma possível?

O tradeoff entre Bias e Variância vai sempre depender do algoritmo a ser utilizado para a construção do modelo. 

Por exemplo:

- O algoritmo de KNN, tem baixo Bias e alta Variância, mas o tradeoff pode ser alterado ao aumentar o valor de k, que aumenta o Bias do modelo.

- O algoritmo de Support Vector Machine (SVM) tem baixo Bias e alta Variância, mas a compensação pode ser alterada aumentando o parâmetro C, que influencia o número de violações da margem permitida nos dados de treinamento, o que aumenta a Viés, mas diminui a Variância.


Nesse sentido, encontrar esse equilíbrio é um processo trabalhoso e iterativo, sendo necessário treinar o modelo diversas vezes com variadas combinações de features, hiperparâmetros e com diferentes conjuntos de dados para treinamento.

## **Validação Cruzada** 

Uma técnica para avaliar a capacidade de generalização do modelo e, consequentemente, lidar da melhor forma com problemas como o overfitting é a validação cruzada (Cross-validation). 

- A validação cruzada é uma técnica que avalia um modelo dividindo os dados em múltiplos subconjuntos, treinando o modelo em um subconjunto e testando-o nos restantes de forma iterativa. Isso tem como intuito avaliar como o modelo se comporta com dados inéditos, simulando uma situação de uso real.

Uma das formas mais conhecidas de aplicar uma validação cruzada em um modelo é por meio da técnica de _k-fold cross-validation_. Nela, o conjunto de dados é dividido de forma aleatória em "K" folds (ou "dobras" em português)

O fluxo metodológico é simples, observe o exemplo abaixo.

```python
# IMPORTS
import pandas as pd
import numpy as np
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import MinMaxScaler, LabelEncoder
from sklearn.model_selection import cross_val_score
import warnings
from sklearn.exceptions import ConvergenceWarning

# =========================================
# 1. GERANDO UM DATASET SINTÉTICO
# =========================================

np.random.seed(42)

# Criando 1000 amostras
n = 1000

dataset = pd.DataFrame({
    'idade': np.random.randint(18, 70, n),
    'renda': np.random.randint(1000, 10000, n),
    'tempo_emprego': np.random.randint(0, 40, n),
    'checking_status': np.random.choice(['baixo', 'médio', 'alto'], n),
    'credit_history': np.random.choice(['bom', 'ruim', 'neutro'], n),
    'purpose': np.random.choice(['carro', 'educação', 'reforma', 'outros'], n),
    'savings_status': np.random.choice(['baixo', 'médio', 'alto'], n),
    'class': np.random.choice(['bom_pagador', 'mau_pagador'], n)
})

# =========================================
# 2. PRÉ-PROCESSAMENTO
# =========================================

# Dummies para variáveis categóricas
dataset_dummies = pd.get_dummies(dataset, 
    columns=['checking_status', 'credit_history', 'purpose', 'savings_status'], 
    drop_first=True
)

# Transformando a variável alvo em numérica
labelencoder = LabelEncoder()
dataset_dummies['class'] = labelencoder.fit_transform(dataset_dummies['class'])

# Separando X e y
X = dataset_dummies.drop(columns=['class'])
y = dataset_dummies['class']

# Normalização
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# =========================================
# 3. DEFINIÇÃO DO MODELO
# =========================================

model = MLPClassifier(hidden_layer_sizes=(25, 25),
                      activation='relu',
                      solver='adam',
                      max_iter=200,
                      random_state=42)

# =========================================
# 4. CROSS-VALIDATION
# =========================================

scores = cross_val_score(model, X_scaled, y, cv=10, scoring='accuracy')

print("Scores individuais:", np.round(scores, 3))
print("Média de acurácia: %.2f%%" % (scores.mean() * 100))
print("Desvio padrão: %.2f%%" % (scores.std() * 100))
```

Nesse exemplo, o conjunto de dados foi dividido em 10 "dobras". 
- No primeiro teste, a dobra 1 é o teste e as dobras 2, 3, 4, 5 são treino.
- No segundo teste, a dobra 2 é o teste e as dobras 1, 3, 4, 5 são treino, e sucessivamente.

Esse processo continua, e as estatísticas são coletadas em cada etapa.

No fim, é retornada uma lista com o resultado de cada um dos testes, para cada um dos subconjuntos, utilizando a métrica padrão do modelo: acurácia, para modelos de classificação, e R2 para modelos de regressão.

### 💡 Dica:

Recomendamos que utilizem para K os valores 5 e 10, pois estes não costumam causar erros de teste altos nem de Bias e nem de Variância.



## **Problemas de classificação**

### **O que é um problema de classificação?**

Um problema de classificação em _machine learning_ é um tipo de aprendizado supervisionado que, como o próprio nome já diz, se concentra em classificar os dados em categorias, classes ou grupos específicos, com base em certas características. Dessa maneira, o objetivo desse tipo de técnica não é prever um valor numérico contínuo, mas sim determinar a qual grupo ou rótulo um item pertence.

As classificações podem ser de dois tipos: binárias ou multiclasse. 
- As binárias, em geral, são de resposta SIM ou NÃO.                 
- As multiclasse são aquelas que possuem várias características.

Como exemplos práticos, temos: 

- Classificação automática de e-mails como "spam" ou "não spam" (classificação binária) 
    
- Identificação do tipo de fruta presente em uma imagem (classificação multiclasse).

### **O que é um classificador?**

Um **classificador** é um modelo que atribui uma **categoria** ou **classe** a cada observação com base nas variáveis de entrada.  
Em outras palavras, enquanto um regressor prevê valores contínuos, o classificador decide **qual grupo ou rótulo** um dado pertence.

Alguns termos equivalentes para a variável alvo em classificação são:

- Rótulo
- Variável dependente
- Classe

Existem diferentes tipos de classificação: **binária** (duas classes), **multiclasse** (mais de duas classes) e **multirrótulo** (uma amostra pode ter várias classes), e todos buscam **aproximar a função** que relaciona as variáveis de entrada às classes corretas:

$$
f(X) \approx y
$$

Aqui, X representa os atributos ou características de cada observação, e y é a **classe prevista** pelo modelo.


### **Métricas de Avaliação para Classificação**

Para explorar as diferentes métricas de avaliação de performance de um modelo preditivo em problemas de classificação, vamos criar um _dataset_ contendo 60 amostras de flores com variáveis categóricas, como cor, tamanho e textura das pétalas, além de uma classe alvo representando o tipo de flor, e rodar um modelo de árvore de decisão.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, LabelEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# -------------------------------
# 1. Criar conjunto de dados categóricos
# -------------------------------
import random

random.seed(42)

cores = ['Vermelha', 'Amarela', 'Azul', 'Branca']
tamanhos = ['Pequena', 'Média', 'Grande']
texturas = ['Lisa', 'Rugosa']

# Gerar 60 amostras aleatórias
data = {
    'Cor_da_petala': [random.choice(cores) for _ in range(60)],
    'Tamanho_da_petala': [random.choice(tamanhos) for _ in range(60)],
    'Textura_da_petala': [random.choice(texturas) for _ in range(60)]
}

df = pd.DataFrame(data)

# Criar o rótulo 
def definir_flor(row):
    if row['Cor_da_petala'] == 'Vermelha' and row['Tamanho_da_petala'] == 'Grande':
        return 'Rosa'
    elif row['Cor_da_petala'] == 'Amarela':
        return 'Girassol'
    elif row['Cor_da_petala'] == 'Azul':
        return 'Íris'
    else:
        return random.choice(['Hibisco', 'Rosa'])

df['Tipo_de_flor'] = df.apply(definir_flor, axis=1)

# -------------------------------
# 2. Definindo variáveis independentes e dependentes
# -------------------------------
X = df[['Cor_da_petala', 'Tamanho_da_petala', 'Textura_da_petala']]
y = df['Tipo_de_flor']

X_encoded = pd.get_dummies(X)
y_encoded = LabelEncoder().fit_transform(y)

X_train, X_test, y_train, y_test = train_test_split(X_encoded, y_encoded, test_size=0.3, random_state=42)

# -------------------------------
# 3. Modelo 
# -------------------------------
model = DecisionTreeClassifier(max_depth=4, random_state=42)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print(df)
```

#### **🎯 Acurácia**

Essa é uma métrica que calcula a proporção de todas as previsões corretas realizadas pelo modelo. 

Ela é calculada como: 

<p align="center">
  
$$
\frac{\text{Verdadeiros Positivos} + \text{Verdadeiros Negativos}}{\text{Total de Previsões}}
$$

</p>


**Quando usar:**

- Quando as classes estão balanceadas (quantidade parecida entre elas).

**Vantagens:**

- Fácil de interpretar.

- Boa visão geral do desempenho.

**Desvantagens:**

Enganosa em dados desbalanceados (ex.: 95% de uma classe → modelo pode acertar 95% só prevendo sempre a mesma).

```python
acc = accuracy_score(y_test, y_pred)

print(f"Acurácia:  {acc:.2f}")
```

#### **🎯 Precisão**

A precisão mede a proporção de previsões positivas que foram corretas. Por exemplo, entre as flores previstas como “Rosa”, quantas realmente são “Rosa”.


<p align="center">
  
$$
\frac{\text{Verdadeiros Positivos}}{\text{Verdadeiros Positivos} + \text{Falsos Positivos}}
$$

</p>

**Quando usar:**

- Quando o custo de um falso positivo é alto

**Vantagens:**

- Mede o quanto das previsões positivas estão corretas.

**Desvantagens:**

- Ignora os falsos negativos.

```python
prec = precision_score(y_test, y_pred, average='weighted', zero_division=0)

print(f"Precisão:  {prec:.2f}")
```

#### **🎯 Recall (ou Sensibilidade)**

A sensibilidade mede a proporção de casos positivos reais que o modelo identificou corretamente. Como exemplo, entre todas as flores “Rosa” reais, quantas o modelo acertou.

<p align="center">
  
$$
\frac{\text{Verdadeiros Positivos}}{\text{Verdadeiros Positivos} + \text{Falsos Negativos}}
$$

</p>

**Quando usar:**

- Quando o custo de um falso negativo é alto.

**Vantagens:**

- Mede a capacidade de encontrar todos os casos positivos.

**Desvantagens:**

- Pode gerar muitos falsos positivos (alta sensibilidade, mas pouca precisão).

```python
rec = recall_score(y_test, y_pred, average='weighted', zero_division=0)
print(f"Revocação: {rec:.2f}")
```

#### **🎯 F1-Score**

Representa uma média harmônica entre precisão e _Recall_, sendo uma maneira de observar em um único número essas duas outras medidas.

<p align="center">
  
$$
\frac{\text {2 * Precisão * Recall}}{\text{Precisão} + \text{Recall}}
$$

</p>

**Quando usar:**

- Quando há desequilíbrio entre classes e você quer equilíbrio entre precisão e recall.

**Vantagens:**

- Combina precisão e recall em um único número.

**Desvantagens:**

- Não mostra qual dos dois (precisão ou recall) está mais fraco.

```python
f1 = f1_score(y_test, y_pred, average='weighted', zero_division=0)
print(f"F1-Score:  {f1:.2f}\n")
```

#### **🎯 Curva ROC-AUC**

A curva ROC mede o quão bom o modelo é para distinguir entre dois rótulos,  plotando a taxa de verdadeiros positivos (sensibilidade) no eixo Y contra a taxa de falsos positivos (1 - especificidade) no eixo X, em diferentes limiares de classificação



AUC é a área sob a curva, sendo derivada da curva ROC, resumindo o desempenho do modelo em um valor de 0 a 1.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelBinarizer
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import roc_curve, auc, roc_auc_score
import matplotlib.pyplot as plt

# 1. Carregar o dataset
iris = load_iris()
X = iris.data
y = iris.target

# 2. Transformar em binário 
y_binary = (y == 0).astype(int)

# 3. Dividir treino e teste
X_train, X_test, y_train, y_test = train_test_split(X, y_binary, test_size=0.3, random_state=42)

# 4. Treinar o modelo
model = GaussianNB()
model.fit(X_train, y_train)

# 5. Obter probabilidades
y_prob = model.predict_proba(X_test)[:, 1]

# 6. Calcular curva ROC e AUC
fpr, tpr, _ = roc_curve(y_test, y_prob)
roc_auc = auc(fpr, tpr)

# 7. Plotar a curva
plt.figure(figsize=(6, 5))
plt.plot(fpr, tpr, color='blue', lw=2, label=f'GaussianNB (AUC = {roc_auc:.3f})')
plt.plot([0, 1], [0, 1], color='gray', lw=1, linestyle='--', label='Aleatório')
plt.xlabel('Taxa de Falsos Positivos (FPR)')
plt.ylabel('Taxa de Verdadeiros Positivos (TPR)')
plt.title('Curva ROC-AUC')
plt.legend(loc='lower right')
plt.grid(True)
plt.show()
```

**Quando usar:**

- Para avaliar classificadores binários.
- Quando há diferentes limiares de decisão e você quer comparar desempenho geral.

**Vantagens:**

- Mostra o trade-off entre sensibilidade e especificidade.
- Útil para comparar modelos independentemente do threshold.

**Desvantagens:**

- Menos informativo em dados muito desbalanceados.
- Pode ser difícil de interpretar.

### **O classificador de Bayes**

O classificador de Bayes é um algoritmo utilizado em problemas de classificação que utiliza o Teorema de Bayes para classificar os dados, sendo amplamente utilizada em tarefas como classificação de texto. 
- O teorema de Bayes calcula a probabilidade de um evento A ocorrer dado que um evento B já ocorreu, atualizando uma previsão ou teoria existente com base em novas evidências.


<p align="center">
  
$$
P(A \mid B) = \frac{P(B \mid A) \cdot P(A)}{P(B)}
$$

</p>

- P(A|B): a probabilidade de A dado B (probabilidade condicional).
- P(B|A): a probabilidade de B dado A.
- P(A): a probabilidade a priori de A.
- P(B): a probabilidade de B.

Existem diversos tipos de classificadores de Bayes, sendo os mais populares:

- Gaussian Naïve Bayes (GaussianNB): usado para variáveis contínuas com distribuição normal, estimando média e desvio padrão de cada classe.

- Multinomial Naïve Bayes (MultinomialNB): indicado para dados discretos (como contagens de palavras), comum em classificação de textos e detecção de spam.

- Bernoulli Naïve Bayes (BernoulliNB): aplicado a variáveis binárias (0/1 ou True/False).


Observe o exemplo abaixo com o conjunto de dados anterior.

```python
from sklearn.naive_bayes import GaussianNB, MultinomialNB, BernoulliNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

modelos = {
    "GaussianNB": GaussianNB(),
    "MultinomialNB": MultinomialNB(),
    "BernoulliNB": BernoulliNB()
}

for nome, modelo in modelos.items():
    modelo.fit(X_train, y_train)
    y_pred = modelo.predict(X_test)
    
    print(f"\n=== {nome} ===")
    print("Acurácia:", round(accuracy_score(y_test, y_pred), 3))
    print("Precisão:", round(precision_score(y_test, y_pred, average='weighted', zero_division=0), 3))
    print("Recall:", round(recall_score(y_test, y_pred, average='weighted'), 3))
    print("F1-Score:", round(f1_score(y_test, y_pred, average='weighted'), 3))
```

## **Problemas de regressão**

### **O que é um problema de regressão?**

Diferentemente dos problemas de classificação, que preveem valores discretos, os problemas de regressão se concentram em prever valores numéricos. Assim como a classificação, a regressão é uma técnica de aprendizado supervisionado, ou seja, utiliza conjuntos de dados com rótulos para treinar algoritmos. 

Neste caso, o modelo aprende com um conjunto de dados rotulados em que para cada exemplo de treinamento é atribuído um valor numérico específico. 

- Assim, ele é treinado para encontrar a relação entre as características dos dados de entrada e seus valores correspondentes. 

- Por fim, ele usa essa relação para prever o valor de novos dados de entrada que ainda não foram rotulados.

- Em outras palavras, é previsto um resultado contínuo (y) com base no valor de uma ou mais variáveis ​​preditoras (x)

Alguns exemplos de problemas de regressão são:
- Prever o preço de uma casa (com base em sua área, número de quartos etc...)
- Estimar o valor do plano de saúde de uma pessoa, considerando a idade do cliente

### **O que é um regressor?**

Neste contexto, um regressor é qualquer variável em nosso modelo que é utilizada para predizer a nossa variável resposta, que é um valor numérico contínuo.

Essa variável também é conhecida por outros nomes, como:

- Variável independente
- Variável preditora
- Variável explicativa

Existem diversos tipos de regressão: linear, polinomial, ridge, lasso, entre outros, e todos têm o mesmo objetivo: aproximar uma função que descreva a relação entre as variáveis preditoras (X) e o valor alvo (y).

Assim, podemos pensar no regressor como uma função matemática que tenta aproximar os dados reais:

$$
f(X) \approx y
$$

X são as variáveis de entrada e y é o valor contínuo que queremos prever.

### **Métricas de Avaliação para Regressão**

A avaliação da performance de um modelo de regressão é dada pela diferença entre os valores reais e os valores preditos, a partir de diferentes métricas.

#### **MSE**

O _Mean Squared Error_ é uma métrica que avalia a média dos quadrados das diferenças entre os valores preditos e os reais em um conjunto de dados, em que:


$$
MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$$

- yi é o valor real;
- y^i é o valor predito;
- n é o número de dados


**Vantagens**:
- É diferenciável, sendo amplamente utilizada em processos de otimização de modelos.

- Penaliza fortemente erros grandes, tornando o modelo mais sensível a desvios significativos.

**Desvantagens**:
- É sensível a outliers, já que os erros são elevados ao quadrado.

- A métrica não está na mesma unidade da variável alvo, o que pode dificultar a interpretação direta.

```python
from sklearn.metrics import mean_squared_error

# Definindo valores reais e valores preditos
y_true = [35, 32, 24, 12, 9]
y_pred = [33.5, 31, 27, 11, 6]

# Calculando MSE
mse = mean_squared_error(y_true, y_pred)
print("Mean Squared Error (MSE):", mse)
```

##### **🔎 Quanto menor o MSE, melhor o modelo.**


#### **RMSE**

O _Root Mean Squared Error_ é a raiz quadrada da média dos quadrados das diferenças entre os valores preditos e os reais. Ele representa a magnitude média do erro em mesma unidade da variável alvo, o que facilita a interpretação dos resultados.

$$
RMSE = \sqrt{\frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2}
$$

- yi é o valor real;
- y^i é o valor predito;
- n é o número de dados

**Vantagens:**

- Está na mesma unidade da variável alvo, o que facilita a interpretação.

- Penaliza erros grandes de forma mais intensa, assim como o MSE.

**Desvantagens:**

- Também é sensível a outliers.

- Pode ser mais difícil de otimizar diretamente, já que envolve uma raiz quadrada.

```python
import numpy as np

# Calculando RMSE
rmse = np.sqrt(mse)
print("Root Mean Squared Error (RMSE):", rmse)
```

#### **MAE**

O _Mean Absolute Error_ mede a média dos valores absolutos das diferenças entre os valores reais e preditos. É uma métrica que indica o erro médio sem dar peso adicional a erros grandes.

$$
MAE = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|
$$

- yi é o valor real;
- y^i é o valor predito;
- n é o número de dados

**Vantagens**:

- Mais robusto a outliers do que o MSE e o RMSE.

- Fácil de interpretar, pois está na mesma unidade da variável alvo.

**Desvantagens**:

- Não diferencia tanto entre erros pequenos e grandes.

- Não é diferenciável em zero, o que pode dificultar algumas otimizações.

```python
from sklearn.metrics import mean_absolute_error

# Calculando MAE
mae = mean_absolute_error(y_true, y_pred)
print("Mean Absolute Error (MAE):", mae)
```

#### **$R^{2}$**

O Coeficiente de Determinação (ou R-squared) mede a proporção da variabilidade da variável dependente que é explicada pelo modelo. O valor de 
varia de 0 a 1, sendo que valores próximos de 1 indicam melhor ajuste.

$$
R^2 = 1 - \frac{\sum_{i=1}^{n} (y_i - \hat{y}_i)^2}{\sum_{i=1}^{n} (y_i - \bar{y})^2}
$$

- yi é o valor real;
- 𝑦^𝑖 é o valor predito; 
- 𝑦 ˉ é a média dos valores reais.

**Vantagens**:

- Fornece uma medida intuitiva de quão bem o modelo explica a variabilidade dos dados.

- É amplamente utilizado e fácil de comparar entre modelos.

**Desvantagens**:

- Pode ser enganoso em modelos não lineares ou com dados desbalanceados.

- Não penaliza a complexidade do modelo


```python
from sklearn.metrics import r2_score

# Calculando R²
r2 = r2_score(y_true, y_pred)
print("R-squared (R²):", r2)
```

Existem outras métricas menos comuns, como MAPE (Mean Absolute Percentage Error), MSLE (Mean Squared Log Error) e MedAE (Median Absolute Error), que podem ser úteis em contextos específicos, por exemplo, quando se quer expressar o erro em termos percentuais.

## **Feature Engineering**

### **O que é Feature Engineering?**

Pode ser definida como a transformação de dados brutos em features (características) mais relevantes, a fim de melhorar o desempenho dos nossos modelos preditivos. Esse processo é uma das principais etapas para a construção de um modelo preditivo eficiente, concentrando-se em maximizar a qualidade dos dados de entrada.



Alguns processos de feature engineering incluem: 

- Transformação de dados de um formato para o outro. 

    - Exemplo: transformar variáveis categóricas em numéricas
    - Exemplo: normalizar e padronizar dados 

- Seleção e limpeza dos dados.

    - Exemplo: remoção de outliers
    - Exemplo: imputação de valores ausentes

- Criação de novas features.

    - Exemplo: juntar dados de latitude e longitude em uma nova variável "coordenada"
    - Exemplo: decompor datas em dia, mês e ano


### **Exemplo prático**

Nesta seção, será demonstrado o processo completo de _Feature Engineering_, uma etapa essencial no pré-processamento de dados para _Machine Learning_.

O fluxo será dividido em três etapas principais:

- Limpeza de dados – remoção de outliers e imputação de valores ausentes;

- Transformação de dados – codificação de variáveis categóricas;

- Criação de novas features – geração de atributos derivados que podem melhorar a performance do modelo.

Para demonstrar como esse processo acontece, vamos trabalhar agora com um conjunto de dados sintético com informações sobre 100 trabalhadores em uma empresa.

```python
import pandas as pd
import numpy as np

np.random.seed(42)

# Criação do dataset
data = {
    'Idade': np.random.randint(18, 70, 100),
    'Salario': np.random.normal(5000, 1500, 100).round(2),
    'Tempo_de_Emprego': np.random.randint(0, 30, 100),
    'Departamento': np.random.choice(['RH', 'TI', 'Financeiro', 'Marketing'], 100),
    'Nivel_Educacao': np.random.choice(['Ensino Médio', 'Graduação', 'Mestrado', 'Doutorado'], 100),
    'Desempenho': np.random.choice(['Baixo', 'Médio', 'Alto'], 100)
}

df = pd.DataFrame(data)

# Valores ausentes
for col in ['Idade', 'Salario', 'Tempo_de_Emprego']:
    df.loc[np.random.choice(df.index, 5, replace=False), col] = np.nan

# Outliers
df.loc[np.random.choice(df.index, 3, replace=False), 'Salario'] = df['Salario'].mean() * 5

print(df.head())
```

### 🧹 **Limpeza de dados**

#### **Remoção de outliers**

```python
Q1 = df['Salario'].quantile(0.25)
Q3 = df['Salario'].quantile(0.75)
IQR = Q3 - Q1

# Limite inferior e superior
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

# Remover outliers
df_sem_outliers = df[(df['Salario'] >= limite_inferior) & (df['Salario'] <= limite_superior)]

print("Antes:", df.shape[0], "linhas")
print("Depois da remoção de outliers:", df_sem_outliers.shape[0], "linhas")
```

##### 💡 Alternativas possíveis:

- Z-score (padronização e remoção de valores fora de 3 desvios-padrão)

- Isolation Forest (método baseado em aprendizado de máquina para detectar anomalias)

#### **Imputação de valores ausentes**

```python
df_limpo = df_sem_outliers.copy()

# Preencher valores ausentes
df_limpo['Idade'].fillna(df_limpo['Idade'].mean(), inplace=True)
df_limpo['Salario'].fillna(df_limpo['Salario'].mean(), inplace=True)
df_limpo['Tempo_de_Emprego'].fillna(df_limpo['Tempo_de_Emprego'].median(), inplace=True)

print("Valores ausentes após a imputação:")
print(df_limpo.isnull().sum())
```

##### 💡 Alternativas possíveis:

- Imputação por moda 

- _KNN Imputer_

- _Iterative Imputer_

### 🔄 **Transformação de dados**

#### **Encoding de Variáveis Categóricas**

Vamos aplicar a técnica de _One-Hot Encoding_ nas variáveis categóricas e _Label Encoding_ na variável alvo.

```python
from sklearn.preprocessing import OneHotEncoder, LabelEncoder

X = df_limpo[['Departamento', 'Nivel_Educacao']]
y = df_limpo['Desempenho']

# One-Hot Encoding nas variáveis preditoras
X_encoded = pd.get_dummies(X, drop_first=True)

# Label Encoding na variável alvo
le = LabelEncoder()
y_encoded = le.fit_transform(y)

X_encoded.head()
```

##### 💡 Alternativas possíveis:

- _Ordinal Encoding_ (para categorias com ordem, ex.: “Baixo”, “Médio”, “Alto”)

- _Target Encoding_ (substitui categorias pela média do alvo — útil em modelos lineares)

- _Binary Encoding_ (boa opção quando há muitas categorias distintas)

### ⚙️ **Criação de novas features**

Aqui são criados atributos derivados que podem capturar relações importantes entre variáveis.

```python
df_features = df_limpo.copy()

# Feature binária: experiência
df_features['Experiente'] = np.where(df_features['Tempo_de_Emprego'] >= 10, 1, 0)

# Feature categórica: faixa etária
df_features['Faixa_Etaria'] = pd.cut(df_features['Idade'], 
                                     bins=[17, 25, 40, 60, 80],
                                     labels=['Jovem', 'Adulto', 'Meia-Idade', 'Sênior'])

# Feature numérica: relação salário/idade
df_features['Salario_por_Idade'] = (df_features['Salario'] / df_features['Idade']).round(2)

df_features[['Idade', 'Tempo_de_Emprego', 'Experiente', 'Faixa_Etaria', 'Salario_por_Idade']].head()
```
